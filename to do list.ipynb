```python
def find_task(tasks, task_id):
    """Return the task with the given ID, or None if it does not exist."""
    for task in tasks:
        if task["id"] == task_id:
            return task
    return None


def add_task(tasks, next_id):
    """Ask for a description and add a new pending task."""
    description = input("Enter task description: ").strip()
    if not description:
        print("Task description cannot be empty.")
        return next_id

    task = {"id": next_id, "description": description, "status": "Pending"}
    tasks.append(task)
    print("Task added successfully!")
    return next_id + 1


def view_tasks(tasks):
    """Display all tasks, or a message when the list is empty."""
    if not tasks:
        print("No tasks found.")
        return

    print("\n===== YOUR TASKS =====")
    for task in tasks:
        print(f"\nID: {task['id']}")
        print(f"Task: {task['description']}")
        print(f"Status: {task['status']}")


def update_task(tasks):
    """Update the description of a task selected by ID."""
    try:
        task_id = int(input("Enter the ID of the task to update: "))
    except ValueError:
        print("Invalid ID. Please enter a whole number.")
        return

    task = find_task(tasks, task_id)
    if task is None:
        print("Task ID not found.")
        return

    description = input("Enter the new task description: ").strip()
    if not description:
        print("Task description cannot be empty.")
        return

    task["description"] = description
    print("Task updated successfully!")


def complete_task(tasks):
    """Mark a pending task as completed."""
    try:
        task_id = int(input("Enter the ID of the task to complete: "))
    except ValueError:
        print("Invalid ID. Please enter a whole number.")
        return

    task = find_task(tasks, task_id)
    if task is None:
        print("Task ID not found.")
    elif task["status"] == "Completed":
        print("This task is already completed.")
    else:
        task["status"] = "Completed"
        print("Task marked as completed!")


def delete_task(tasks):
    """Delete a task selected by ID."""
    try:
        task_id = int(input("Enter the ID of the task to delete: "))
    except ValueError:
        print("Invalid ID. Please enter a whole number.")
        return

    task = find_task(tasks, task_id)
    if task is None:
        print("Task ID not found.")
    else:
        tasks.remove(task)
        print("Task deleted successfully!")


def main():
    """Run the TODO application's menu until the user chooses to exit."""
    tasks = []
    next_id = 1

    while True:
        print("\n===== TODO APPLICATION =====")
        print("1. Add Task")
        print("2. View Tasks")
        print("3. Update Task")
        print("4. Mark Task as Completed")
        print("5. Delete Task")
        print("6. Exit")

        choice = input("Enter your choice: ").strip()

        if choice == "1":
            next_id = add_task(tasks, next_id)
        elif choice == "2":
            view_tasks(tasks)
        elif choice == "3":
            update_task(tasks)
        elif choice == "4":
            complete_task(tasks)
        elif choice == "5":
            delete_task(tasks)
        elif choice == "6":
            print("Goodbye!")
            break
        else:
            print("Invalid choice. Please select an option from 1 to 6.")


if __name__ == "__main__":
    main()
```
